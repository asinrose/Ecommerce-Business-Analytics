```python
## 3. Duplicate Transaction Investigation

Objective:
Identify exact duplicate transaction lines and determine
whether they should be removed from the analytical dataset.
```

```python
# Extract duplicate records
duplicate_rows = df[
    df.duplicated(keep=False)
].sort_values(
    by=["Invoice", "StockCode", "InvoiceDate"]
)

print("Duplicate records including original rows:",
      len(duplicate_rows))

duplicate_rows.head(15)
```

```python
duplicate_groups = (
    df.groupby(list(df.columns), dropna=False)
      .size()
      .reset_index(name="Occurrences")
)

duplicate_groups = duplicate_groups[
    duplicate_groups["Occurrences"] > 1
]

print("Number of duplicate groups:",
      len(duplicate_groups))

duplicate_groups["Occurrences"].value_counts().sort_index()
```

```python
# Compare overlapping transaction conditions

cancelled_mask = df["Invoice"].astype(str).str.startswith("C")
negative_mask = df["Quantity"] < 0
invalid_price_mask = df["Price"] <= 0

print("Cancelled AND negative quantity:",
      (cancelled_mask & negative_mask).sum())

print("Negative quantity AND invalid price:",
      (negative_mask & invalid_price_mask).sum())

print("Cancelled AND invalid price:",
      (cancelled_mask & invalid_price_mask).sum())

print("All three conditions:",
      (cancelled_mask & negative_mask & invalid_price_mask).sum())
```

```python
# Check whether missing descriptions have identifiable product codes

missing_codes = df.loc[
    df["Description"].isna(),
    "StockCode"
].unique()

print("Unique stock codes with missing descriptions:",
      len(missing_codes))
```

```python
description_lookup = (
    df.dropna(subset=["Description"])
      .groupby("StockCode")["Description"]
      .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
)

print("Missing description records:",
      df["Description"].isna().sum())

print("Missing descriptions with a recoverable stock code:",
      df.loc[
          df["Description"].isna(),
          "StockCode"
      ].isin(description_lookup.index).sum())
```

```python
## 4. Data Cleaning and Transformation

Objective:
Clean transaction records, standardize data,
handle missing descriptions and separate sales,
returns and cancellations.
```

```python
# Preserve the original dataset
df_raw = df.copy()

# Remove exact duplicate rows, keeping the first occurrence
df_clean = df_raw.drop_duplicates().copy()

print("Original records:", len(df_raw))
print("Records after deduplication:", len(df_clean))
print("Removed duplicate records:",
      len(df_raw) - len(df_clean))
```

```python
df_clean.columns = [
    "InvoiceNo",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "UnitPrice",
    "CustomerID",
    "Country"
]

print(df_clean.columns.tolist())
```

```python
# Build product description lookup
description_lookup = (
    df_clean.dropna(subset=["Description"])
    .groupby("StockCode")["Description"]
    .agg(lambda x: x.mode().iloc[0])
)

# Fill missing descriptions using matching StockCode
missing_before = df_clean["Description"].isna().sum()

df_clean["Description"] = (
    df_clean["Description"]
    .fillna(df_clean["StockCode"].map(description_lookup))
)

missing_after = df_clean["Description"].isna().sum()

print("Missing descriptions before:", missing_before)
print("Missing descriptions after:", missing_after)
print("Descriptions recovered:",
      missing_before - missing_after)
```

```python
df_clean["CustomerID"] = (
    pd.to_numeric(df_clean["CustomerID"], errors="coerce")
    .astype("Int64")
)

print(df_clean["CustomerID"].dtype)
```

```python
cancelled_mask = (
    df_clean["InvoiceNo"].astype(str).str.startswith("C")
)

cancellations = df_clean[cancelled_mask].copy()

print("Cancellation records:", len(cancellations))
```

```python
returns = df_clean[
    (~cancelled_mask) &
    (df_clean["Quantity"] < 0)
].copy()

print("Non-cancelled negative quantity records:",
      len(returns))
```

```python
sales = df_clean[
    (~cancelled_mask) &
    (df_clean["Quantity"] > 0) &
    (df_clean["UnitPrice"] > 0)
].copy()

# Calculate transaction-line revenue
sales["Revenue"] = (
    sales["Quantity"] * sales["UnitPrice"]
)

print("Sales records:", len(sales))
print("Total revenue: £",
      round(sales["Revenue"].sum(), 2))
print("Total units sold:", sales["Quantity"].sum())
```

```python
invalid_price_records = df_clean[
    df_clean["UnitPrice"] <= 0
].copy()

print("Invalid price records:",
      len(invalid_price_records))
```

```python
print("ORIGINAL DATASET:", df_raw.shape)
print("CLEANED DATASET:", df_clean.shape)

print("\nSALES DATASET:", sales.shape)
print("RETURNS DATASET:", returns.shape)
print("CANCELLATIONS DATASET:", cancellations.shape)
print("INVALID PRICE DATASET:", invalid_price_records.shape)

print("\nMissing values in sales:")
print(sales.isnull().sum())
```
